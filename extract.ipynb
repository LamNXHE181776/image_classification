```python
from classify.data import dataset

data = dataset.ClassificationDataModule(
    data_dir= r"datasets/dataset_2",
    batch_size=32,
    num_workers=2,
    )

train_dict, val_dict = data.setup()
```

```python
list(train_dict.items())
```

```python
df = pd.DataFrame(list(train_dict.items()), columns=['carCompany', 'count']).explode('count')
df_val = pd.DataFrame(list(val_dict.items()), columns=['carCompany', 'count']).explode('count')
```

```python
df
```

```python
df_count1 = df.groupby('carCompany')['count'].count().reset_index().sort_values(by='carCompany', ascending=True)
df_count1.columns = ['carCompany', 'count_train']
df_count2 = df_val.groupby('carCompany')['count'].count().reset_index().sort_values(by='carCompany', ascending=True)
df_count2.columns = ['carCompany', 'count_val']
df_count = pd.merge(df_count1, df_count2, on='carCompany', how='outer')
df_count.columns = ['Các lớp', 'Số lượng mẫu trong tập train ngày 15/6/2026', 'Số lượng mẫu trong tập val ngày 15/6/2026']
df_count = df_count.replace({
    "ALFA_ROMEO": "ALFA ROMEO",
    "CHEYRY": "CHERY",
    "LYNK&CO": "LYNK & CO",
    "MERCEDES-BENZ-BENZ-BENZ": "MERCEDES-BENZ"
})
df_count
```

```python
df2 = pd.read_excel("/home/lam/Documents/Book 1.xlsx")
df2["Các lớp"] = df2["Các lớp"].replace({
    "ALFA_ROMEO": "ALFA ROMEO",
    "CHERRY": "CHERY",
    "LYNK&CO": "LYNK & CO",
    "MERCEDES": "MERCEDES-BENZ"
})
df2
```

```python
df3 = pd.merge(df2, df_count, how="outer")
df3.fillna(0, inplace=True)
df3.to_excel("/home/lam/Documents/summary.xlsx", index=False)
df3
```
